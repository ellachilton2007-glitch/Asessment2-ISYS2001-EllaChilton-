# **Personal Spending Assistant**

# Step 1: Understand the Problem

The purpose of this project is to create a personal spending assistant that helps users understand their financial transactions. Users will be able to upload a CSV file containing their spending information and receive summaries about their expenses. The assistant will calculate total spending, show spending by category, identify the category with the highest spending and answer questions about the uploaded data.

# Step 2: Identify Inputs and Outputs

Inputs
- A CSV file containing financial transactions.
- Transaction descriptions, categories and amounts.
- A question typed by the user.

Outputs
- Total spending.
- Spending totals for each category.
- The category with the highest spending.
- A response to the user's question based on their financial data.
- Error messages if the uploaded file or information is invalid.

# Step 3: Work Through an Example by Hand
Suppose a user has four financial transactions:
- Supermarket: Groceries, $60

- Bus card: Transport, $20

- Food shop: Groceries, $40

- Cinema: Entertainment, $30

To calculate the total spending I add all four amounts together:

60 + 20 + 40 + 30 = $150

Then I calculate the total of each category:

- Groceries = $100

- Transport = $20

- Entertainment = $30

The category with the highest total is Groceries, with $100. This helps me understand as it shows what the program should should calculate, before I start coding.

# Step 5: Python Implementation

```python
import pandas as pd

# Example financial transactions
transactions = {
    "Description": ["Supermarket", "Bus card", "Food shop", "Cinema"],
    "Category": ["Groceries", "Transport", "Groceries", "Entertainment"],
    "Amount": [60, 20, 40, 30]
}

# Convert the transactions into a pandas DataFrame
df = pd.DataFrame(transactions)

# Display the transactions
print(df)
```

```python
# Calculate the total spending
total_spending = df["Amount"].sum()

print("Total spending: $", total_spending)

# Calculate spending for each category
category_spending = df.groupby("Category")["Amount"].sum()

print("\nSpending by category:")
print(category_spending.to_string())

# Find the category with the highest spending
biggest_category = category_spending.idxmax()
biggest_amount = category_spending.max()

print("\nCategory with the highest spending:")
print(biggest_category, "- $", biggest_amount)

def check_transactions(df):
    # Check that the data is not empty
    if df.empty:
        return "Error: No transactions found."

    # Check that the required columns exist
    required_columns = ["Description", "Category", "Amount"]

    for column in required_columns:
        if column not in df.columns:
            return "Error: Missing required column: " + column

    # Check that descriptions and categories are not missing
    if df["Description"].isna().any() or df["Category"].isna().any():
        return "Error: Descriptions and categories cannot be empty."

    # Check that the amounts are numbers
    if not pd.api.types.is_numeric_dtype(df["Amount"]):
        return "Error: Amounts must be numbers."

    # Check that amounts are not missing
    if df["Amount"].isna().any():
        return "Error: Amounts cannot be empty."

    # Check that amounts are not negative
    if (df["Amount"] < 0).any():
        return "Error: Amounts cannot be negative."

    return "Transaction data looks good!"
```

```python
def spending_summary(df):
    total_spending = df["Amount"].sum()
    category_spending = df.groupby("Category")["Amount"].sum()
    biggest_category = category_spending.idxmax()
    biggest_amount = category_spending.max()

    return total_spending, category_spending, biggest_category, biggest_amount
```

# Create Gradio Interface

```python
import gradio as gr

def greet_user(name):
    if name.strip() == "":
        return "Please enter your name."
    return "Hello, " + name + "! Welcome to your Personal Spending Assistant."

demo = gr.Interface(
    fn=greet_user,
    inputs=gr.Textbox(label="Enter your name"),
    outputs=gr.Textbox(label="Welcome message"),
    title="Personal Spending Assistant",
    description="Your tool for understanding your spending."
)

demo.launch(share=True)
```

```python
def analyse_csv(file):
    if file is None:
        return "Please upload a CSV file."

    try:
        data = pd.read_csv(file.name)

        # Check that the uploaded data is valid
        result = check_transactions(data)

        if result != "Transaction data looks good!":
            return result

        # Calculate the spending summary
        total, categories, biggest, amount = spending_summary(data)

        summary = f"Total spending: ${total:.2f}\n\n"
        summary += "Spending by category:\n"
        summary += categories.to_string()
        summary += f"\n\nBiggest category: {biggest} - ${amount:.2f}"

        return summary

    except Exception as error:
        return f"Could not read the file. Please upload a valid CSV. Details: {error}"


demo = gr.Interface(
    fn=analyse_csv,
    inputs=gr.File(label="Upload your transaction CSV", file_types=[".csv"], type="filepath"),
    outputs=gr.Textbox(label="Your spending summary", lines=8),
    title="Personal Spending Assistant",
    description="Upload your transaction CSV to analyse your spending."
)

demo.launch(share=True)
```

# Add Gemini Assistance

```python
!pip install -q google-genai
```

```python
from google import genai
from google.colab import userdata

API_KEY = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=API_KEY)

print("Gemini connection is ready!")
```

```python
prompt = "In one sentence, explain why tracking personal spending can be useful."

response = client.models.generate_content(
    model="gemini-flash-latest",
    contents=prompt
)

print(response.text)
```

```python
def answer_spending_question(file, question):
    if file is None:
        return "Please upload your transaction CSV first."

    if question.strip() == "":
        return "Please enter a question about your spending."

    try:
        data = pd.read_csv(file.name)

        result = check_transactions(data)
        if result != "Transaction data looks good!":
            return result

        total, categories, biggest, amount = spending_summary(data)

        spending_information = f"""
        Total spending: ${total:.2f}

        Spending by category:
        {categories.to_string()}

        Biggest spending category: {biggest}
        Amount in biggest category: ${amount:.2f}
        """

        prompt = f"""
        You are a helpful personal spending assistant.

        Use only the spending information provided below to answer
        the user's question. Do not make up financial information.
        If the information is not enough to answer, say so.

        Spending information:
        {spending_information}

        User's question:
        {question}
        """

        response = client.models.generate_content(
            model="gemini-flash-latest",
            contents=prompt
        )

        return response.text

    except Exception as error:
        return "Something went wrong. Please check your CSV and try again."


demo = gr.Interface(
    fn=answer_spending_question,
    inputs=[
        gr.File(
            label="Upload your transaction CSV",
            file_types=[".csv"],
            type="filepath"
        ),
        gr.Textbox(
            label="Ask a question about your spending",
            placeholder="Which category did I spend the most on?"
        )
    ],
    outputs=gr.Textbox(
        label="Gemini's answer",
        lines=8
    ),
    title="Personal Spending Assistant",
    description="Upload your transactions and ask questions about your spending."
)

demo.launch(share=True)
```
